# Scratch MLP Gradient Checking

```python
import sys
import os
import numpy as np

sys.path.append(os.path.abspath('..')) 

from src.data_loader import load_svhn_data
from src.features import extract_flattened
from src.mlp import MLPFromScratch

print("--- Loading Debug Dataset ---")
X_train_raw, y_train, _, _ = load_svhn_data(
    data_dir="../data/raw", 
    n_train=15000, 
    n_test=3000, 
    random_state=42, 
    normalize=True,
    balanced=True
)

print("\n--- Extracting Flattened Features ---")
X_train_flat, _ = extract_flattened(X_train_raw, X_train_raw, standardize=True)

# We take 3 images to do gradient checking
X_debug = X_train_flat[:3]
y_debug = y_train[:3]

print(f"\nDebug features shape: {X_debug.shape}")
print(f"Debug labels shape: {y_debug.shape}")
```

```python
# Initialize a lightweight network
print("--- Initializing MLP for Gradient Checking ---")
model_debug = MLPFromScratch(
    input_size=X_debug.shape[1], 
    hidden_layers=[128, 64], 
    num_classes=10, 
    init_method="Xavier"
)

# Gradient checking for three weights per layer
model_debug.gradient_checking(X_debug, y_debug, epsilon=1e-7, max_params_to_check=3)
```
